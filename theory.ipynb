## Pre-collapse burning stages

A star above $\sim 10$–$11\,M_\odot$ burns through elements from $\mathrm{H}$ to a $\mathrm{Fe}$ core, each stage ignited when the star contract and increase temperature enough until the energy released increases enough to balance energy losses and gravitational attraction. The contraction and heating results in the nuclei becoming energetic enough to either tunnel through their mutual electrostatic repulsion to fuse (lighter elements than $\mathrm{Si}$), or thermal photons becoming energetic enough to knock out an $\alpha$, $p$ or $n$, and then re-absorbed for a net fusion into heavier species ($\mathrm{Ne}$ and heavier elements than $\mathrm{Si}$ inclusive) [Woosley and Heger, 2015](https://arxiv.org/abs/1505.06712). 

Stage durations shorten steeply with mass, with broad stages shown below:

Central values from Table I (pp. 1019–1020); reactions and ashes from §III.A (p. 1026), §III.E (pp. 1029–1031) and §IV.A (pp. 1032–1035) of [Woosley and Heger 2002](https://journals.aps.org/rmp/abstract/10.1103/RevModPhys.74.1015).

- Only weak reactions change the electron fraction $Y_e = \sum_i Z_i X_i / A_i$; strong and electromagnetic reactions conserve $Z$ and $A$ separately. 
- The neutron excess $\eta = 1 - 2Y_e$ starts at $\approx 0.002\,(Z/Z_\odot)$, begins rising during $\mathrm{C}$-burning, and reaches $\eta \gtrsim 0.01$ in central $\mathrm{O}$-burning ($Y_e \approx 0.495$), "neutronising" the core before $\mathrm{Si}$ ignites. 

- From $\mathrm{C}$-burning onwards, energy leaves the core as free-streaming $\nu$ rather than diffused $\gamma$ in a dense stellar interior. 
- A major channel for thermal loss is pair annihilation, where with volumetric emissivity rising $\propto T^9$, a contracting and heating core loses energy fater and must burn faster to compensate [Itoh et al. 1996](https://ui.adsabs.harvard.edu/abs/1996ApJS..102..411I/abstract). 

The complete $\mathrm{Si}$-burning networks exhibit high temperatures with fastest rates, thus requiring shortest timesteps as well for accurate stellar-evolution modelling.

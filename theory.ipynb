# Star's burning bit too hot

A star above $\sim 10$–$11\,M_\odot$ burns through elements from $\mathrm{H}$ to a $\mathrm{Fe}$ core, each stage ignited when the star contracts and increases temperature enough until the energy released increases enough to balance energy losses, and radiative pressure balance gravitational attraction. The contraction and heating results in the nuclei becoming energetic enough to either tunnel through their mutual electrostatic repulsion to fuse (lighter elements than $\mathrm{Si}$), or thermal photons becoming energetic enough to knock out an $\alpha$, $p$ or $n$, and then re-absorbed for a net fusion into heavier species ($\mathrm{Ne}$ and heavier elements than $\mathrm{Si}$ inclusive) [Woosley and Heger, 2015](https://arxiv.org/abs/1505.06712), [Woosley et al. 2002](https://journals.aps.org/rmp/abstract/10.1103/RevModPhys.74.1015). 

Stage durations shorten steeply with mass, with broad stages shown below:

- Only weak reactions change the electron fraction $Y_e = \sum_i Z_i X_i / A_i$; strong and electromagnetic reactions both conserve $Z$ and $A$. 
- The neutron excess $\eta = 1 - 2Y_e$ starts at $\approx 0.002\,(Z/Z_\odot)$ after $\mathrm{He}$-burning, begins rising during $\mathrm{C}$-burning, and reaches $\eta \gtrsim 0.01$ in central $\mathrm{O}$-burning ($Y_e \approx 0.495$), "neutronising" the core before $\mathrm{Si}$ ignites. 
- From $\mathrm{C}$-burning onwards, energy leaves the core as free-streaming $\nu$ rather than diffused $\gamma$ in a dense stellar interior [Itoh et al. 1996](https://ui.adsabs.harvard.edu/abs/1996ApJS..102..411I/abstract). 

- The complete $\mathrm{Si}$-burning networks exhibit high temperatures with fastest rates, thus requiring shortest timesteps as well for accurate stellar-evolution modelling. 

# Star's gonna go boom boom boom

At the end of $\mathrm{Si}$-burning, the core is an $\mathrm{Fe}$ sphere supported by electron degeneracy pressure. 

### Microphysics

Consider electrons being placed in a box of volume $V$ with periodic boundary conditions ($\mathbf{k} = \tfrac{2\pi\mathbf{n}}{L}$ for integer $\mathbf{n}$). This forms a lattice with one state per cell of $k$-volume $(\tfrac{2\pi}{L})^3$. The number of states $dN$ in $d^3k$ is

$$
dN = \frac{V\,d^3k}{(2\pi)^3} = \frac{V\,d^3p}{(2\pi)^3\hbar^3} = \frac{V\,d^3p}{h^3}
$$

where $\mathbf{p} = \hbar\mathbf{k}$.

Since an $e^-$ is spin-$\tfrac{1}{2}$, each spatial state $\mathbf{p}$ can have either spin $m_s = \pm\tfrac12$, introducing a degeneracy factor $g = 2$. ($\gamma$ will also have a degeneracy of $2$ from its two polarisations). The number of available states in a phase-space element is therefore,

$$
dN_{\rm states} = \frac{2}{h^3}\,d^3x\,d^3p
$$

Filling from the ground energy up at $T = 0$, a new $e^-$ will fill the lowest-energy state right above a filled sphere of radius $p_F$ in momentum space. By integrating over the sphere per unit volume, we can determine the number density of $e^-$, $n_e$,

$$
n_e = \int_{|p|<p_F} \frac{dN_{\rm states}}{d^3x} = \frac{2}{h^3}\int_0^{p_F} 4\pi p^2\,dp = \frac{8\pi}{3h^3}\,p_F^3
\qquad\Longrightarrow\qquad
p_F = \hbar\,(3\pi^2 n_e)^{1/3}
$$

For a fixed, finite volume $h^3$ in phase space, adding $e^-$ and "squeezing" in the position space (increasing $n_e$) will "spread" out momentum space (increasing $p_F$) which manifests as a degeneracy pressure. 

Next, consider pressure as the momentum flux across a surface. Starting from the momentum flux tensor of a non-interacting gas,

$$
\Pi_{ij} = \int n(\mathbf{p})\,p_i v_j\,d^3p
$$

where $n(\mathbf{p})$ is the number density per unit momentum-space volume.

For an isotropic gas, $\Pi_{ij} = P\,\delta_{ij}$ and

$$
3P = \int n(\mathbf{p})\,p\,v(p)\,d^3p
\qquad\Longrightarrow\qquad
P = \frac{1}{3}\int n(\mathbf{p})\,p\,v(p)\,d^3p
$$

where $n(\mathbf{p}) = \tfrac{2}{h^3}$ and which over the same filled sphere is

$$
P = \frac{4\pi}{3}\frac{2}{h^3}\int_0^{p_F} p^3\,v(p)\,dp = \frac{8\pi}{3h^3}\int_0^{p_F} p^3\,v(p)\,dp
$$

In the non-relativistic limit, $v(p) = p/m_e$,

$$
P = \frac{8\pi}{3h^3}\int_0^{p_F} p^3\,\frac{p}{m_e}\,dp = \frac{8\pi}{15h^3m_e}\,p_F^5
$$

For the typical conditions of the regimes of interest, at $\rho = 10^{8}\,\mathrm{g\,cm^{-3}}$ and $Y_e = 0.5$, we have $p_Fc = 1.90\,\mathrm{MeV}$ with rest energy $m_ec^2 = 0.511\,\mathrm{MeV}$ (evaluated with $n_e = \rho N_A Y_e$, [established below](#eqn-electron-number-density)). Since $x = \tfrac{p_Fc}{m_ec^2} = 3.7 > 1$, we thus take the relativistic limit $v(p) \to c$:

$$
P = \frac{8\pi c}{3h^3}\int_0^{p_F} p^3\,dp = \frac{2\pi c}{3h^3}\,p_F^4 = \frac{(3\pi^2)^{1/3}}{4}\,\hbar c\,n_e^{4/3}
$$

### Nuclear physics

For a nuclear species $i$, let $Z_i$ be its proton number, $A_i$ be its nucleon number, $n_i$ be its number density ($\mathrm{cm^{-3}}$), $X_i$ be its mass fraction of total baryonic mass such that $\sum_i X_i = 1$, and $Y_i$ be its abundance per baryon given by $Y_i \equiv \tfrac{n_i}{n_B} \equiv \tfrac{X_i}{A_i}$.

The plasma is charge-neutral, so $e^-$ balance the $p$ and

$$
n_e = \sum_i Z_i n_i
\qquad\Longrightarrow\qquad
Y_e = \sum_i Z_i Y_i = \sum_i \frac{Z_i X_i}{A_i}
$$

where each nucleon has mass $m_u \approx \frac{1\,\mathrm{g\,mol^{-1}}}{N_A}$

$$
\rho = n_B m_u
\qquad\Longrightarrow\qquad
n_e = n_B Y_e = \frac{\rho Y_e}{m_u}
$$

<a id="eqn-electron-number-density"></a>

$$
P = \frac{(3\pi^2)^{1/3}}{4}\,\hbar c\,\left(\frac{\rho Y_e}{m_u}\right)^{4/3}
$$

where written in the form $P = K\rho^{(n + 1)/n}$, we have

$$
P = K\rho^{4/3},
\qquad
K = \frac{(3\pi^2)^{1/3}}{4}\,\hbar c\,\left(\frac{Y_e}{m_u}\right)^{4/3}
$$

with $n = 3$.

### Macrophysics

Consider a hydrostatic equilibrium where the electron degeneracy pressure and gravitational attraction is in parity

$$
\frac{dP}{dr} = -\frac{Gm(r)\rho}{r^2},
$$

where we take $\tfrac{dm}{dr} = 4\pi r^2\rho$ such that

$$
\frac{1}{r^2}\frac{d}{dr}\!\left(\frac{r^2}{\rho}\frac{dP}{dr}\right) = -4\pi G\rho
$$

Now, we consider the form $P = K\rho^{(n+1)/n}$ from earlier

$$
\frac{dP}{dr} = K\frac{n+1}{n}\rho^{1/n}\frac{d\rho}{dr}
\qquad\Longrightarrow\qquad
\frac{1}{\rho}\frac{dP}{dr} = K\frac{n+1}{n}\rho^{1/n-1}\frac{d\rho}{dr} = (n+1)K\,\frac{d}{dr}\!\left(\rho^{1/n}\right)
$$

Consider $\rho = \rho_c\theta^n$ such that $\rho^{1/n}$ is linear in $\theta$. In terms of the dimensionless parameter $\theta$,

$$
\frac{1}{\rho}\frac{dP}{dr} = (n+1)K\rho_c^{1/n}\frac{d\theta}{dr}
\qquad\Longrightarrow\qquad
\frac{dP}{dr} = (n+1)K\rho_c^{1/n}\rho\frac{d\theta}{dr}
$$

Back into the hydrostatic equilibrium,

$$
(n+1)K\rho_c^{1/n}\,\frac{1}{r^2}\frac{d}{dr}\!\left(r^2\frac{d\theta}{dr}\right) = -4\pi G\rho_c\theta^n
\qquad\Longrightarrow\qquad
\left[\frac{(n+1)K}{4\pi G}\rho_c^{(1-n)/n}\right]\frac{1}{r^2}\frac{d}{dr}\!\left(r^2\frac{d\theta}{dr}\right) = -\theta^n
$$

where we define

$$
a^2 \equiv \frac{(n+1)K}{4\pi G}\rho_c^{(1-n)/n}
\quad
[\mathrm{cm^2}]
$$

and consider another $r = a\xi$ where

$$
\frac{1}{r^2}\frac{d}{dr}\!\left(r^2\frac{d\theta}{dr}\right) = \frac{1}{a^2}\cdot\frac{1}{\xi^2}\frac{d}{d\xi}\!\left(\xi^2\frac{d\theta}{d\xi}\right)
$$

such that in terms of the dimensionless parameter $\xi$,

$$
a^2\cdot\frac{1}{a^2}\cdot\frac{1}{\xi^2}\frac{d}{d\xi}\!\left(\xi^2\frac{d\theta}{d\xi}\right) = \boxed{\frac{1}{\xi^2}\frac{d}{d\xi}\!\left(\xi^2\frac{d\theta}{d\xi}\right) = -\theta^n}
$$

At the boundary conditions,
- $\theta(0)=1$ recovers $\rho(0)=\rho_c$
- $\tfrac{d\theta}{d\xi}(0)=0$ because near the centre $m\approx\frac43\pi r^3\rho_c$, so $\tfrac{dP}{dr}\propto -r\to0$, avoiding any density cusps
- at the first root $\xi_1$ where $\theta = 0$, is the surface $R = a\xi_1$ where pressure vanishes

The mass of an enclosed volume of the star is

$$
M = \int_0^R 4\pi r^2\rho\,dr = 4\pi a^3\rho_c\int_0^{\xi_1}\xi^2\theta^n\,d\xi
$$

in terms of $\theta$ and $\xi$. Since $\xi^2\theta^n = -\tfrac{d}{d\xi}(\xi^2\cdot\tfrac{d\theta}{d\xi})$,

$$
M = -4\pi a^3\rho_c\bigg[\xi^2\cdot\frac{d\theta}{d\xi}\bigg]_0^{\xi_1} = -4\pi a^3\rho_c\xi_1^2\frac{d\theta}{d\xi}(\xi_1) = 4\pi a^3\rho_c\xi_1^2\bigg|\frac{d\theta}{d\xi}(\xi_1)\bigg|
$$

where $\frac{d\theta}{d\xi}(0) = 0$ and $\frac{d\theta}{d\xi}(\xi_1) < 0$ since density decreases outwards

Since

$$
a^3 = \left[\frac{(n+1)K}{4\pi G}\right]^{3/2}\rho_c^{3(1-n)/2n},
\qquad\Longrightarrow\qquad
M = 4\pi\left[\frac{(n+1)K}{4\pi G}\right]^{3/2}\rho_c^{(3-n)/2n}\,\xi_1^2\bigg|\frac{d\theta}{d\xi}(\xi_1)\bigg|
$$

### Microphysics + Macrophysics = (Micro + Macro)Physics = Physics

For $n = 3$,

$$
M_{\rm Ch} = 4\pi\left(\frac{K}{\pi G}\right)^{3/2}\,\xi_1^2\left|\frac{d\theta}{d\xi}(\xi_1)\right|\approx 5.8237\,Y_e^{2}\;M_\odot
$$

where $K = (1.2435\times10^{15})Y_e^{4/3}\,\mathrm{g^{-1/3}\,cm^{3}\,s^{-2}}$ and the Lane-Emden equation is solved numerically to yield $\xi_1^2\left|\tfrac{d\theta}{d\xi}(\xi_1)\right| \approx 2.01824$.

- At $Y_e \approx 0.5$, $M_{\rm Ch} \approx 1.456\,M_\odot$

$M_{\rm Ch}$ describes the critical mass where electron degeneracy pressure can no longer support the star. For $\mathrm{O}$-burning networks, the $A\approx30$-$37$ ash which recaptures $^{16}\mathrm{O} + ^{16}\mathrm{O}$ ejecta ($p$, $n$, $\alpha$) is susceptible to weak reactions; for $\mathrm{Si}$-burning networks, similar recapture of photodisintegration products ($p$, $n$, $\alpha$) and $\mathrm{Fe}$-group are susceptible to weak reactions. Both networks' weak reactions produce free-streaming $\nu$ which escape the core, preventing the reverse reaction from occurring. Effectively, this reduces $Y_e$, thus quadratically reducing $M_{\rm Ch}$ as well until eventually beyond which electron degeneracy pressure can no longer support the star and the star starts to collapse. [Woosley et al. 2002](https://journals.aps.org/rmp/abstract/10.1103/RevModPhys.74.1015)

# More thermodynamics

A thermodynamic description of a $\mathrm{Si}$-burning core can start from a decomposition of its pressure

$$
P = \underbrace{n_ikT}_{= P_{\rm ion}} + \underbrace{\frac{(3\pi^2)^{1/3}}{4}\,\hbar c\,\left(\frac{\rho Y_e}{m_u}\right)^{4/3}}_{= P_e} + \underbrace{\frac{aT^4}{3}}_{= P_{\rm rad}} + P_{\rm Coulomb}
$$

where $n_i = \tfrac{\rho N_A}{\bar{A}}$ and $a = 7.566 \times 10^{-15}\,\mathrm{erg\,cm^{-3}\,K^{-4}}$ [Mohr et al. 2022](https://arxiv.org/abs/2409.03787).

- $P_{\rm ion}$ is the ideal gas pressure
- $P_e$ is described by degenerate and relativistic $e$
- $P_{\rm rad}$

$$
P_e = \frac{8\pi}{3h^3}\int_0^\infty \frac{p^3 v(p)}{\exp\left(\frac{E(p)-\mu}{kT}\right)+1},dp,
\qquad
v(p) = \frac{pc^2}{E(p)},
\qquad
E = \sqrt{p^2c^2 + m_e^2c^4}
$$

# Star's burning bit too hot

A star above $\sim 10$–$11\,M_\odot$ burns through elements from $\mathrm{H}$ to a $\mathrm{Fe}$ core, each stage ignited when the star contracts and increases temperature enough until the energy released increases enough to balance energy losses, and radiative pressure balance gravitational attraction. The contraction and heating results in the nuclei becoming energetic enough to either tunnel through their mutual electrostatic repulsion to fuse (lighter elements than $\mathrm{Si}$), or thermal photons becoming energetic enough to knock out an $\alpha$, $p$ or $n$, and then re-absorbed for a net fusion into heavier species ($\mathrm{Ne}$ and heavier elements than $\mathrm{Si}$ inclusive) [Woosley and Heger, 2015](https://arxiv.org/abs/1505.06712) and [Woosley et al. 2002](https://journals.aps.org/rmp/abstract/10.1103/RevModPhys.74.1015). 

Stage durations shorten steeply with mass, with broad stages shown below:

- From $\mathrm{C}$-burning onwards, energy leaves the core as free-streaming $\nu$ rather than diffused $\gamma$ in a dense stellar interior [Itoh et al. 1996](https://ui.adsabs.harvard.edu/abs/1996ApJS..102..411I/abstract). 

The complete $\mathrm{Si}$-burning networks exhibit high temperatures with fastest rates, thus requiring shortest timesteps as well for accurate stellar-evolution modelling.

### Microphysics

Consider electrons being placed in a box of volume $V$ with periodic boundary conditions ($\mathbf{k} = \tfrac{2\pi\mathbf{n}}{L}$ for integer $\mathbf{n}$). This forms a lattice with one state per cell of $k$-volume $\left(\tfrac{2\pi}{L}\right)^3$. The number of states $dN$ in $d^3k$ is

$$
dN = \frac{V\,d^3k}{(2\pi)^3} = \frac{V\,d^3p}{(2\pi)^3\hbar^3} = \frac{V\,d^3p}{h^3}
$$

where $\mathbf{p} = \hbar\mathbf{k}$.

Since an $e^-$ is spin-$\tfrac{1}{2}$, each spatial state $\mathbf{p}$ can have either spin $m_s = \pm\tfrac12$, introducing a degeneracy factor $g = 2$. The number of available states in a phase-space element is therefore,

$$
dN_{\rm states} = \frac{2}{h^3}\,d^3x\,d^3p
$$

Now, how many of these available states are filled by $e^-$?

Consider a system free to exchange both energy and particles with a thermal reservoir where they are both isolated. We take a microstate of the composite as a pair $(i,\,r)$ with the system in its microstate $i$ and the reservoir in its microstate $r$ such that

$$
E_i + E_r = E_{\rm tot},
\qquad
N_i + N_r = N_{\rm tot}
$$

From the postulate of equal a priori probabilities applied to the composite, where all states have the same energy and number of particles $E_{\rm tot}$ and $N_{\rm tot}$ respectively,

$$
P(i,\,r) = \frac{1}{\Omega_{\rm tot}}
$$

For a microstate $i$ of the system of energy $E_i$ with $N_i$ particles, we have

$$
P_i = \sum_r P(i,r) = \frac{\Omega_R(E_r,\,N_r)}{\Omega_{\rm tot}} \propto \Omega_R(E_{\rm tot} - E_i,\,N_{\rm tot} - N_i)
$$

where $\Omega_R$ is the number of microstates accessible to a reservoir with energy $E_r$ and number of particles $N_r$. 

For a reservoir large enough such that $E_i \ll E_{\rm tot}$ and $N_i \ll N_{\rm tot}$, working in logarithms and taking to the first order term of the Taylor expansion, 

$$
\ln\Omega_R(E_{\rm tot}-E_i,\,N_{\rm tot}-N_i) \simeq \ln\Omega_R(E_{\rm tot},\,N_{\rm tot}) - E_i\left(\frac{\partial\ln\Omega_R}{\partial E}\right)_{V,N} - N_i\left(\frac{\partial\ln\Omega_R}{\partial N}\right)_{E,V}
$$

where both derivatives are evaluated at $(E_{\rm tot},\,N_{\rm tot})$. Since for the reservoir, $S = k\,\ln\Omega_R$ and $dE = T\,dS - P\,dV + \mu\,dN$,

$$
\left(\frac{\partial\ln\Omega_R}{\partial E}\right)_{V,N} = \frac{1}{k}\left(\frac{\partial S}{\partial E}\right)_{V,N} = \frac{1}{k}\frac{1}{T} \equiv \beta,
\qquad
\left(\frac{\partial\ln\Omega_R}{\partial N}\right)_{E,V} = \frac{1}{k}\left(\frac{\partial S}{\partial N}\right)_{E,V} = \frac{1}{k}\left(-\frac{\mu}{T}\right) = -\beta\mu
$$

where $T$ and $\mu$ are of the reservoir, and thus the system's at equilibrium. Then,

$$
\ln\Omega_R(E_{\rm tot} - E_i,\,N_{\rm tot} - N_i) \simeq \ln\Omega_R(E_{\rm tot},\,N_{\rm tot}) - E_i\,\beta + N_i\,\beta\mu = \ln\Omega_R(E_{\rm tot},\,N_{\rm tot}) - \beta\,\left(E_i - N_i\,\mu\right)
$$

$$
P_i \propto \Omega_R(E_{\rm tot},N_{\rm tot})\,e^{-\beta(E_i-\mu N_i)} \propto e^{-\beta(E_i-\mu N_i)}
\qquad\Longrightarrow\qquad
P_i = C\,e^{-\beta(E_i-\mu N_i)}
$$

where we introduce a proportionality constant $C$ irrespective of the energy $E_i$ or number of particles $N_i$ of each microstate, and sum over every microstate the system can occupy,

$$
\sum_i P_i = 1 = C\,\sum_i e^{-\beta(E_i-\mu N_i)}
$$

so that can define the grand canonical partition function $\mathcal{Z}$ to write

$$
P_i = \frac{e^{-\beta(E_i-\mu N_i)}}{\mathcal{Z}},
\qquad
\mathcal{Z}\equiv \sum_i e^{-\beta(E_i-\mu N_i)}
$$

Suppose instead now the energy $E_i$ and number of particles $N_i$ in a system is allowed to fluctuate. Consider $M$ independent copies of the system in an ensemble where each microstate $i$ has probability $P_i$ and expect $M_i = M\,P_i$ copies as $M \to \infty$. The number of possible permutations of assigning $M$ copies of the system to the microstates is $\frac{M!}{\prod_iM_i!}$ and the ensemble entropy is

$$
S_{\rm ensemble} = k\ln\left(\frac{M!}{\prod_iM_i!}\right)
$$

where with Stirling's approximation $\ln M! \simeq M\ln M - M$ and $\sum_i M_i = M$,

$$
\ln\left(\frac{M!}{\prod_iM_i!}\right) \simeq \ln M! - \ln\prod_iM_i! = \left(M\ln M - M\right) - \sum_i\left(M_i\ln M_i - M_i\right) = M\ln M - \sum_i M_i\ln M_i
$$

and substituting $M_i = MP_i$,

$$
\sum_i M_i\ln M_i = M\sum_i P_i\left(\ln M + \ln P_i\right) = M\ln M + M\sum_i P_i\ln P_i
$$

and thus

$$
\ln\left(\frac{M!}{\prod_iM_i!}\right) \simeq -M\sum_i P_i\ln P_i
$$

which yields the entropy per copy

$$
S = \frac{S_{\rm ensemble}}{M} = -k\sum_i P_i\ln P_i
$$

As a sanity check, taking $P_i = \tfrac{1}{\Omega}$ where the system has a definite energy $E_i$ and number of particles $N_i$ such that $P_i$ is equal over all microstates $i$ the system is in, we recover

$$
S = -k\sum_{i=1}^{\Omega}\frac{1}{\Omega}\ln\frac{1}{\Omega} = k\ln\Omega
$$

Continuing from the entropy per copy,

$$
S = -k\sum_i P_i\ln P_i = -k\sum_i P_i\ln\frac{e^{-\beta(E_i-\mu N_i)}}{\mathcal{Z}} = -k\sum_i P_i\left(-\beta\left(E_i - \mu N_i\right) - \ln\mathcal{Z}\right)
$$

$$
S = k\beta\sum_i P_i\left(E_i - \mu N_i\right) + k\ln\mathcal{Z}\sum_i P_i
$$

where since $\sum_i P_i E_i = \langle E\rangle$, $\sum_i P_i N_i = \langle N\rangle$ and $\sum_i P_i = 1$,

$$
S = \frac{\langle E\rangle - \mu\langle N\rangle}{T} + k\ln\mathcal{Z}
\qquad\Longrightarrow\qquad
-kT\ln\mathcal{Z} = \langle E\rangle - TS - \mu\langle N\rangle \equiv \Omega
$$

arriving at the definition for the grand potential, $\Omega$. Suppose the system is extensive, then Euler's theorem gives $E = TS - PV + \mu N$ and

$$
\Omega = -kT\ln\mathcal{Z} = -PV
$$

We can also consider in terms of volume densities, where $\sigma = \tfrac{S}{V}$, $u = \tfrac{E}{V}$ and $n = \tfrac{N}{V}$, we have

$$
\langle E\rangle - TS - \mu\langle N\rangle = -PV
\qquad\Longrightarrow\qquad
T\sigma = u + P - \mu n
$$

Since a microstate $i$ specifies the occupation numbers $\{n_s\}$ over each state $\{s\}$. Assuming **(why can we assume so, why can Coloumbic interactions be deferred later)** a non-interacting gas, the total energy $E_i$ and number of particles $N_i$ of a microstate $i$ is

$$
E_i = \sum_sE_sn_s,
\qquad
N_i = \sum_sn_s
\qquad\Longrightarrow\qquad
E_i - \mu N_i = \sum_s E_s n_s - \mu\sum_s n_s = \sum_s\left(E_s - \mu\right)n_s
$$

$$
\mathcal{Z} = \sum_i e^{-\beta(E_i - \mu N_i)} = \sum_i e^{-\beta\sum_s\left(E_s - \mu\right)n_s} = \sum_i \prod_s e^{-\beta(E_s-\mu)n_s}
$$

where since a microstate $i$ is the collection $\{n_s\}$, the sum over microstates is a product of the sums over each configuration where

$$
\sum_i = \sum_{\{n_s\}} = \sum_{n_1}\sum_{n_2}\sum_{n_3}\cdots
$$

and thus,

$$
\mathcal{Z} = \sum_{\{n_s\}}\prod_s e^{-\beta(E_s-\mu)n_s} = \prod_s\sum_{n_s} e^{-\beta(E_s-\mu)n_s} \equiv \prod_s \mathcal{Z}_s
$$

For a single state $s$ of an $e^-$ with energy $E_s$, Pauli exclusion principle restricts $n_s \in \{0,\,1\}$ and thus

$$
\mathcal{Z}_s = \sum_{n_s = 0}^{1} e^{-\beta(E_s - \mu)n_s}
= \underbrace{1}_{n_s\,=\,0} + \underbrace{e^{-\beta(E_s-\mu)}}_{n_s\,=\,1},
\qquad
\beta \equiv \frac{1}{kT}
$$

Now, we consider the probability of a microstate having occupancies $\{n_s\}$

$$
P_{\{n_s\}} = \frac{e^{-\beta(E_i-\mu N_i)}}{\mathcal{Z}} = \frac{e^{-\beta\left(\sum_s(E_s-\mu)n_s\right)}}{\prod_s\mathcal{Z}_s} = \frac{\prod_s e^{-\beta(E_s-\mu)n_s}}{\prod_s\mathcal{Z}_s} = \prod_s\frac{e^{-\beta(E_s-\mu)n_s}}{\mathcal{Z}_s}
$$

which allows us to define the probability of a state $s$ having an occupancy $n_s$

$$
\prod_s P_s(n_s) \equiv \prod_s\frac{e^{-\beta(E_s-\mu)n_s}}{\mathcal{Z}_s}
\qquad\Longrightarrow\qquad
P_s(n_s) = \frac{e^{-\beta(E_s-\mu)n_s}}{\mathcal{Z}_s}
$$

enumerated over $n_s \in \{0,\,1\}$,

$$
P_s(0) = \frac{1}{1 + e^{-\beta(E_s-\mu)}},
\qquad
P_s(1) = \frac{e^{-\beta(E_s-\mu)}}{1 + e^{-\beta(E_s-\mu)}},
\qquad
P_s(0) + P_s(1) = 1
$$

and the mean occupancy $\langle n_s\rangle$ is thus,

$$
\langle n_s\rangle = 0\cdot P_s(0) + 1\cdot P_s(1) = P_s(1) \equiv f(E)
\qquad\Longrightarrow\qquad
f(E) = \frac{1}{e^{(E-\mu)/kT}+1}
$$

The occupied phase-space density is then $n(\mathbf{p}) = \tfrac{2}{h^3}f(p)$, and the number density of $e^-$ becomes

$$
n_e = \frac{2}{h^3}\int_0^\infty 4\pi p^2 f(p)\,dp = \frac{8\pi}{h^3}\int_0^\infty \frac{p^2\,dp}{\exp\left(\frac{E(p)-\mu}{kT}\right)+1},
\qquad
E(p) = \sqrt{p^2c^2 + m_e^2c^4}
$$

Next, consider pressure as the momentum flux across a surface. Starting from the momentum flux tensor of a non-interacting gas,

$$
\Pi_{ij} = \int n(\mathbf{p})\,p_i v_j\,d^3p
$$

where $n(\mathbf{p})$ is the number density per unit momentum-space volume.

For an isotropic gas, $\Pi_{ij} = P\,\delta_{ij}$ and

$$
3P = \int n(\mathbf{p})\,p\,v(p)\,d^3p
\qquad\Longrightarrow\qquad
P = \frac{1}{3}\int n(\mathbf{p})\,p\,v(p)\,d^3p
$$

where with $n(\mathbf{p}) = \tfrac{2}{h^3}f(p)$,

$$
P_e = \frac{8\pi}{3h^3}\int_0^\infty \frac{p^3\,v(p)}{\exp\left(\frac{E(p)-\mu}{kT}\right)+1}\,dp,
\qquad
v(p) = \frac{pc^2}{E(p)}
$$

### Nuclear physics

For a nuclear species $i$, let $Z_i$ be its proton number, $A_i$ be its nucleon number, $n_i$ be its number density ($\mathrm{cm^{-3}}$), $X_i$ be its mass fraction of total baryonic mass such that $\sum_i X_i = 1$, and $Y_i$ be its abundance per baryon given by $Y_i \equiv \tfrac{n_i}{n_B} \equiv \tfrac{X_i}{A_i}$.

The plasma is charge-neutral, so $e^-$ balance the $p$ and

$$
n_e = \sum_i Z_i n_i
\qquad\Longrightarrow\qquad
Y_e = \sum_i Z_i Y_i = \sum_i \frac{Z_i X_i}{A_i}
$$

where each nucleon has mass $m_u \approx \frac{1\,\mathrm{g\,mol^{-1}}}{N_A}$

$$
\rho = n_B m_u
\qquad\Longrightarrow\qquad
n_e = n_B Y_e = \frac{\rho Y_e}{m_u}
\qquad\Longrightarrow\qquad
\frac{8\pi}{h^3}\int_0^\infty \frac{p^2\,dp}{\exp\left(\frac{E(p)-\mu}{kT}\right)+1} = \frac{\rho Y_e}{m_u}
$$

### Macrophysics

Consider a hydrostatic equilibrium where the electron degeneracy pressure and gravitational attraction is in parity

$$
\frac{dP}{dr} = -\frac{Gm(r)\rho}{r^2},
$$

where we take $\tfrac{dm}{dr} = 4\pi r^2\rho$ such that

$$
\frac{1}{r^2}\frac{d}{dr}\!\left(\frac{r^2}{\rho}\frac{dP}{dr}\right) = -4\pi G\rho
$$

Now, we consider the form $P = K\rho^{(n+1)/n}$ from earlier

$$
\frac{dP}{dr} = K\frac{n+1}{n}\rho^{1/n}\frac{d\rho}{dr}
\qquad\Longrightarrow\qquad
\frac{1}{\rho}\frac{dP}{dr} = K\frac{n+1}{n}\rho^{1/n-1}\frac{d\rho}{dr} = (n+1)K\,\frac{d}{dr}\!\left(\rho^{1/n}\right)
$$

Consider $\rho \equiv \rho_c\theta^n$ such that $\rho^{1/n}$ is linear in $\theta$. In terms of the dimensionless parameter $\theta$,

$$
\frac{1}{\rho}\frac{dP}{dr} = (n+1)K\rho_c^{1/n}\frac{d\theta}{dr}
\qquad\Longrightarrow\qquad
\frac{dP}{dr} = (n+1)K\rho_c^{1/n}\rho\frac{d\theta}{dr}
$$

Back into the hydrostatic equilibrium,

$$
(n+1)K\rho_c^{1/n}\,\frac{1}{r^2}\frac{d}{dr}\!\left(r^2\frac{d\theta}{dr}\right) = -4\pi G\rho_c\theta^n
\qquad\Longrightarrow\qquad
\left[\frac{(n+1)K}{4\pi G}\rho_c^{(1-n)/n}\right]\frac{1}{r^2}\frac{d}{dr}\!\left(r^2\frac{d\theta}{dr}\right) = -\theta^n
$$

where we define

$$
a^2 \equiv \frac{(n+1)K}{4\pi G}\rho_c^{(1-n)/n}
\quad
[\mathrm{cm^2}]
$$

and consider another $r \equiv a\xi$ where

$$
\frac{1}{r^2}\frac{d}{dr}\!\left(r^2\frac{d\theta}{dr}\right) = \frac{1}{a^2}\cdot\frac{1}{\xi^2}\frac{d}{d\xi}\!\left(\xi^2\frac{d\theta}{d\xi}\right)
$$

such that in terms of the dimensionless parameter $\xi$,

$$
a^2\cdot\frac{1}{a^2}\cdot\frac{1}{\xi^2}\frac{d}{d\xi}\!\left(\xi^2\frac{d\theta}{d\xi}\right) = \boxed{\frac{1}{\xi^2}\frac{d}{d\xi}\!\left(\xi^2\frac{d\theta}{d\xi}\right) = -\theta^n}
$$

At the boundary conditions,
- $\theta(0)=1$ recovers $\rho(0)=\rho_c$
- $\tfrac{d\theta}{d\xi}(0)=0$ because near the centre $m\approx\frac43\pi r^3\rho_c$, so $\tfrac{dP}{dr}\propto -r\to0$, avoiding any density cusps
- at the first root $\xi_1$ where $\theta = 0$, is the surface $R = a\xi_1$ where pressure vanishes

The mass of an enclosed volume of the star is

$$
M = \int_0^R 4\pi r^2\rho\,dr = 4\pi a^3\rho_c\int_0^{\xi_1}\xi^2\theta^n\,d\xi
$$

in terms of $\theta$ and $\xi$. Since $\xi^2\theta^n = -\tfrac{d}{d\xi}(\xi^2\cdot\tfrac{d\theta}{d\xi})$,

$$
M = -4\pi a^3\rho_c\bigg[\xi^2\cdot\frac{d\theta}{d\xi}\bigg]_0^{\xi_1} = -4\pi a^3\rho_c\xi_1^2\frac{d\theta}{d\xi}(\xi_1) = 4\pi a^3\rho_c\xi_1^2\bigg|\frac{d\theta}{d\xi}(\xi_1)\bigg|
$$

where $\frac{d\theta}{d\xi}(0) = 0$ and $\frac{d\theta}{d\xi}(\xi_1) < 0$ since density decreases outwards

Since

$$
a^3 = \left[\frac{(n+1)K}{4\pi G}\right]^{3/2}\rho_c^{3(1-n)/2n},
\qquad\Longrightarrow\qquad
M = 4\pi\left[\frac{(n+1)K}{4\pi G}\right]^{3/2}\rho_c^{(3-n)/2n}\,\xi_1^2\bigg|\frac{d\theta}{d\xi}(\xi_1)\bigg|
$$

### Microphysics + Macrophysics = (Micro + Macro)Physics = Physics

For $n = 3$,

$$
M_{\rm Ch} = 4\pi\left(\frac{K}{\pi G}\right)^{3/2}\,\xi_1^2\left|\frac{d\theta}{d\xi}(\xi_1)\right|\approx 5.8237\,Y_e^{2}\;M_\odot
$$

where $K = (1.2435\times10^{15})Y_e^{4/3}\,\mathrm{g^{-1/3}\,cm^{3}\,s^{-2}}$ and the Lane-Emden equation is solved numerically to yield $\xi_1^2\left|\tfrac{d\theta}{d\xi}(\xi_1)\right| \approx 2.01824$.

- At $Y_e \approx 0.5$, $M_{\rm Ch} \approx 1.456\,M_\odot$

$M_{\rm Ch}$ describes the critical mass where electron degeneracy pressure can no longer support the star. For $\mathrm{O}$-burning networks, the $A\approx30$-$37$ ash which recaptures $^{16}\mathrm{O} + ^{16}\mathrm{O}$ ejecta ($p$, $n$, $\alpha$) is susceptible to weak reactions; for $\mathrm{Si}$-burning networks, similar recapture of photodisintegration products ($p$, $n$, $\alpha$) and $\mathrm{Fe}$-group are susceptible to weak reactions. Both networks' weak reactions produce free-streaming $\nu$ which escape the core, preventing the reverse reaction from occurring. Effectively, this reduces $Y_e$, thus quadratically reducing $M_{\rm Ch}$ as well until eventually beyond which electron degeneracy pressure can no longer support the star and the star starts to collapse. [Woosley et al. 2002](https://journals.aps.org/rmp/abstract/10.1103/RevModPhys.74.1015)

# More thermodynamics

A thermodynamic description of a $\mathrm{Si}$-burning core can start from a decomposition of its pressure

$$
P = \underbrace{n_ikT}_{= P_{\rm ion}} + \underbrace{\frac{(3\pi^2)^{1/3}}{4}\,\hbar c\,\left(\frac{\rho Y_e}{m_u}\right)^{4/3}}_{= P_e} + \underbrace{\frac{aT^4}{3}}_{= P_{\rm rad}} + P_{\rm Coulomb}
$$

where $n_i = \tfrac{\rho N_A}{\bar{A}}$ and $a = 7.566 \times 10^{-15}\,\mathrm{erg\,cm^{-3}\,K^{-4}}$ [Mohr et al. 2022](https://arxiv.org/abs/2409.03787).

- $P_{\rm ion}$ is the ideal gas pressure
- $P_e$ is described by degenerate and relativistic $e$
- $P_{\rm rad}$

$$
P_e = \frac{8\pi}{3h^3}\int_0^\infty \frac{p^3 v(p)}{\exp\left(\frac{E(p)-\mu}{kT}\right)+1},dp,
\qquad
v(p) = \frac{pc^2}{E(p)},
\qquad
E = \sqrt{p^2c^2 + m_e^2c^4}
$$
